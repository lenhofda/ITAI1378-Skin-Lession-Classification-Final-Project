## Data Cleaning: Missing Values, Duplicates & Data Types - Kei'Chara Fleeks

This section fixes the data quality issues found in the EDA. Cleaning is done on copies of the training and testing sets, and `train_data` / `test_data` are replaced with the cleaned versions at the end so every later section uses clean data.

- **Missing values:** fixed on both sets (the rule doesn't learn anything from the data, so there's no leakage).
- **Duplicates:** retained. The test set is left untouched so it still reflects real, unseen data.
- **Data types:** checked and corrected on both sets.


```python
# Step 0: Keep untouched copies so we can show before/after samples
train_before = train_data.copy()
test_before = test_data.copy()

train_clean_dc = train_data.copy()
test_clean_dc = test_data.copy()

print("Training shape before cleaning:", train_before.shape)
print("Testing shape before cleaning: ", test_before.shape)
```

### Step 1: NaN / Null Values

The EDA found 7 missing `TotalCharges` values in the training set, and every one of them has `tenure = 0`. These are brand new customers who haven't been billed yet, so the correct value is **0**, not the mean or median.

Some versions of this dataset store blank `TotalCharges` as a space (`" "`) instead of NaN, so the column is also converted with `pd.to_numeric(errors="coerce")` to catch hidden blanks.


```python
# Step 1a: Count missing values BEFORE (including blank strings hidden as text)
def count_missing(df):
    blanks = df.apply(lambda col: col.astype(str).str.strip().eq("").sum())
    return pd.DataFrame({"NaN": df.isnull().sum(), "Blank strings": blanks}).query("NaN > 0 or `Blank strings` > 0")

print("Training set - missing values BEFORE:")
print(count_missing(train_clean_dc))
print("\nTesting set - missing values BEFORE:")
print(count_missing(test_clean_dc))

# Step 1b: Show the affected rows BEFORE
missing_idx = train_clean_dc.index[pd.to_numeric(train_clean_dc["TotalCharges"], errors="coerce").isnull()]
print("\nAffected training rows BEFORE:")
print(train_clean_dc.loc[missing_idx, ["tenure", "MonthlyCharges", "TotalCharges", "Churn"]])

# Step 1c: Confirm every missing TotalCharges has tenure = 0 (justifies filling with 0)
for name, df in [("train", train_clean_dc), ("test", test_clean_dc)]:
    tc = pd.to_numeric(df["TotalCharges"], errors="coerce")
    print(f"{name}: all missing TotalCharges have tenure = 0 ->", (df.loc[tc.isnull(), "tenure"] == 0).all())
```

```python
# Step 1d: Fix - convert to numeric and fill missing TotalCharges with 0
for df in (train_clean_dc, test_clean_dc):
    df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)

# Step 1e: Show the same rows AFTER and confirm no missing values remain
print("Affected training rows AFTER:")
print(train_clean_dc.loc[missing_idx, ["tenure", "MonthlyCharges", "TotalCharges", "Churn"]])

print("\nTotal missing values AFTER - train:", train_clean_dc.isnull().sum().sum(),
      "| test:", test_clean_dc.isnull().sum().sum())
```

### Step 3: Data Types

Each column should have a type that matches what it represents:

| Column(s) | Target type | Reason |
|---|---|---|
| `tenure` | int | Whole number of months |
| `MonthlyCharges`, `TotalCharges` | float | Money amounts |
| `SeniorCitizen` | int (0/1) | Already a binary flag |
| All other columns | string, whitespace stripped | Categories, encoded in the next section |

Text values are also stripped of extra spaces so `"Yes "` and `"Yes"` aren't treated as different categories.


```python
# Step 3a: Data types BEFORE
dtypes_before = train_before.dtypes.astype(str)

# Step 3b: Fix data types on both sets
numeric_float = ["MonthlyCharges", "TotalCharges"]
numeric_int = ["tenure", "SeniorCitizen"]

for df in (train_clean_dc, test_clean_dc):
    for col in numeric_float:
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("float64")
    for col in numeric_int:
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("int64")
    text_cols = [c for c in df.columns if c not in numeric_float + numeric_int]
    for col in text_cols:
        df[col] = df[col].astype(str).str.strip()

# Step 3c: Data types BEFORE vs AFTER side by side
dtype_compare = pd.DataFrame({"Before": dtypes_before, "After": train_clean_dc.dtypes.astype(str)})
dtype_compare["Changed"] = dtype_compare["Before"] != dtype_compare["After"]
print("Data types BEFORE vs AFTER (training set):")
print(dtype_compare)

# Step 3d: Confirm category labels are clean (no stray spaces or unexpected values)
print("\nUnique values per text column AFTER:")
for col in [c for c in train_clean_dc.columns if c not in numeric_float + numeric_int]:
    print(f"  {col}: {sorted(train_clean_dc[col].unique())}")
```

## Categorical Encoding + One-Hot Encoding + Feature Engineering - Aaliyah Phillips

```python
import pandas as pd
import matplotlib.pyplot as plt

# Make a working copy of the cleaned training data
train_clean = train_data.copy()

# Display categorical columns
print("Categorical Columns: Before")
cat_cols_before = train_clean.select_dtypes(exclude='number').columns.tolist()
print(cat_cols_before)

print("\nSample Data Before Encoding & Engineering:")
print(train_clean[['gender', 'Partner', 'Contract', 'InternetService', 'PaymentMethod']].head())
```

```python
# Step 1: Feature Engineering
# Feature 1: Charges per Month of Tenure
print("\nEngineering Features")

# Feature 1: Charges per Month of Tenure
train_clean['ChargesPerMonth'] = train_clean['TotalCharges'] / (train_clean['tenure'] + 1)

# Feature 2: Tenure Cohort / Lifecycle Group
# Grouping tenure: 0-12m (new), 13-24m (settled), 25-48m (established), 49-72m (loyal)
tenure_bins = [-1, 12, 24, 48, 72]
tenure_labels = ['0-12m', '13-24m', '25-48m', '49-72m']
train_clean['TenureGroup'] = pd.cut(train_clean['tenure'], bins=tenure_bins, labels=tenure_labels)

# Feature 3: Long-Term Customer Indicator (tenure >= 24 months)
train_clean['LongTermCustomer'] = (train_clean['tenure'] >= 24).astype(int)

# Feature 4: Protection & Support Service Count
support_services = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport']
train_clean['SupportServiceCount'] = train_clean[support_services].apply(
    lambda row: (row == 'Yes').sum(), axis=1
)

print("Sample of Engineered Features:")
print(train_clean[['tenure', 'TenureGroup', 'LongTermCustomer', 'ChargesPerMonth', 'SupportServiceCount']].head())
```

```python
# Step 2: Categorical Encoding (Binary & Ordinal)
print("\n--- Applying Categorical & Ordinal Encoding ---")

# 1. Target Variable (Churn)
if 'Churn' in train_clean.columns:
    train_clean['Churn'] = train_clean['Churn'].astype(str).str.strip().map({'Yes': 1, 'No': 0})

# 2. Binary Features (Yes / No)
binary_cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']
for col in binary_cols:
    if col in train_clean.columns:
        train_clean[col] = train_clean[col].astype(str).str.strip().map({'Yes': 1, 'No': 0})

# 3. Gender (Female: 0, Male: 1)
if 'gender' in train_clean.columns:
    train_clean['gender'] = train_clean['gender'].astype(str).str.strip().map({'Female': 0, 'Male': 1})

# 4. Ordinal Encoding: Contract
if 'Contract' in train_clean.columns:
    contract_mapping = {'Month-to-month': 0, 'One year': 1, 'Two year': 2}
    train_clean['Contract_Ordinal'] = train_clean['Contract'].astype(str).str.strip().map(contract_mapping)
    contract_before_plot = train_clean['Contract'].copy()
    train_clean = train_clean.drop(columns=['Contract'])
else:
    contract_before_plot = None

# 5. Drop customerID if present
if 'customerID' in train_clean.columns:
    train_clean = train_clean.drop(columns=['customerID'])
```

```python
# Step 3: One-Hot Encoding
print("\n Applying One-Hot Encoding ")

nominal_cols = ['MultipleLines', 'InternetService', 'OnlineSecurity',
    'OnlineBackup', 'DeviceProtection', 'TechSupport','StreamingTV',
    'StreamingMovies', 'PaymentMethod', 'TenureGroup'
]
# Only encode nominal columns that exist in DataFrame
nominal_cols = [col for col in nominal_cols if col in train_clean.columns]

train_processed = pd.get_dummies(train_clean, columns=nominal_cols, drop_first=True, dtype=int)

print(f"Shape Before: {train_data.shape}")
print(f"Shape After:  {train_processed.shape}")
```

```python
# Step 4: Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Visualization 1: Before & After Contract Encoding
if contract_before_plot is not None and 'Contract_Ordinal' in train_processed.columns:
    contract_before_plot.value_counts().plot(
        kind='bar', ax=axes[0], color='steelblue', edgecolor='black'
    )
    axes[0].set_title("Before Encoding: Raw Contract")
    axes[0].set_xlabel("Contract Type")
    axes[0].set_ylabel("Count")
    axes[0].tick_params(axis='x', rotation=15)

    train_processed['Contract_Ordinal'].value_counts().sort_index().plot(
        kind='bar', ax=axes[1], color='coral', edgecolor='black'
    )
    axes[1].set_title("After Encoding: Contract_Ordinal (0, 1, 2)")
    axes[1].set_xlabel("Ordinal Value")
    axes[1].set_ylabel("Count")
    axes[1].tick_params(axis='x', rotation=0)

plt.show()
```

```python
## Class Balancing
# Separate features from the target
X = train_scaled.drop('Churn', axis=1)
y = train_scaled['Churn']

# Show class counts before balancing
print("Class distribution before balancing:")
print(y.value_counts())

# Plot class distribution before balancing
plt.figure(figsize=(6, 4))
y.value_counts().sort_index().plot(kind='bar')
plt.title('Class Distribution Before Balancing')
plt.xlabel('Churn (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.show()

# Balance the classes using Random Oversampling
from imblearn.over_sampling import RandomOverSampler

ros = RandomOverSampler(random_state=42)
X_balanced, y_balanced = ros.fit_resample(X, y)

# Show class counts after balancing
print("\nClass distribution after balancing:")
print(y_balanced.value_counts())

# Plot class distribution after balancing
plt.figure(figsize=(6, 4))
y_balanced.value_counts().sort_index().plot(kind='bar')
plt.title('Class Distribution After Balancing')
plt.xlabel('Churn (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.show()

# Combine the balanced features and target into one dataset
train_balanced = X_balanced.copy()
train_balanced['Churn'] = y_balanced

print("\nBalanced training data shape:", train_balanced.shape)
print(train_balanced.head())
```

The original training dataset contained 3,622 customers who did not churn (Churn = 0) and 1,308 customers who did churn (Churn = 1), showing an imbalance between the two classes.

To address this imbalance, Random Oversampling was applied to the minority class (Churn = 1) after Standard Scaling was completed. This ensured that the features were scaled using the original training data before balancing the classes.

After balancing, both classes contained 3,622 observations, resulting in a balanced training dataset of 7,244 rows and 36 columns.

Matplotlib was used to visualize the class distribution before and after balancing. The graphs demonstrate that Random Oversampling successfully balanced the two classes.

Only the training data was oversampled. The test dataset was left unchanged to allow unbiased model evaluation.
